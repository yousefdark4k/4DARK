# 📖 Quick Guide: What is Chain Upscaling?

If you're new to this, all these dropdowns probably look a bit overwhelming. Don't worry—it's actually super simple!

---

### ⛓️ What is a "Chain"?
Instead of just blowing up a video and making it look like pixelated garbage, a **Chain** lets you run your file through multiple processing steps in a row (e.g., *shrink it ➡️ clean it ➡️ sharpen it ➡️ upscale it*).

> ⚠️ **Important:** The chain executes sequentially from **top to bottom**.

---

### 🎛️ The Toolkit

| Tool / Model | What it Does | Best Used For / Why? |
| :--- | :--- | :--- |
| **Percentages (25% - 80%)** | Downscales and resizes the video/image first. | Squashes ugly compression artifacts and heavily speeds up the final upscale. |
| **Proteus V3 (Soft)** | Heavily restores original quality. | Delivers **maximum detail**. Excellent for standalone images and real-life (IRL) videos. |
| **Proteus V3 (Strong)** | Topaz-style aggressive processing. | Great for Anime-IRL hybrid videos. *(Caution: Might denoise a bit too much).* |
| **Faster Proteus** | Lightweight, high-speed model. | Use as a secondary upscale model or for inputs that are already high quality. |
| **Sharpen (5% - 50%)** | Recovers lost details. | Use this **before** upscaling to prep the line work/edges. |
| **Force_1080p** | Locks the final output resolution to exactly 1080p. | Hard-caps the output size no matter what crazy math happened earlier in your chain. |

---

### 💡 Pro-Tips & Recommendations

* **Test Small First:** Grab a single test image and experiment with different combinations. Once you find the perfect settings, commit and upscale your entire video.
* **Supported Formats:** This pipeline fully supports **Videos**, **Images**, **Video Folders**, and **Video Images (Image Sequences)**.
